```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def L(mu, x):
    return x**2 + mu*(1 - x)

x = np.arange(-5, 5, 0.1)

# Q1, (a)

plt.figure(figsize=(10, 6))

plt.plot(x, L(0, x), label=r'$\mu = 0$', color='blue')
plt.plot(0, L(0, 0), marker='o', color='blue')

plt.plot(x, L(1, x), label=r'$\mu = 1$', color='green')
plt.plot(0.5, L(1, 0.5), marker='o', color='green')

plt.plot(x, L(2, x), label=r'$\mu = 2$', color='red')
plt.plot(1, L(2, 1), marker='o', color='red')

plt.plot(x, L(3, x), label=r'$\mu = 3$', color='magenta')
plt.plot(1.5, L(3, 1.5), marker='o', color='magenta')

plt.xlabel(r'$x$')
plt.ylabel(r'$\mathcal{L}\left(\mu, x\right)$')
plt.ylim(-1, 15)
plt.legend()
plt.grid()

plt.show()
```

```python
def d(mu):
    return mu - (mu**2)/4

# Q1, (b)

plt.figure(figsize=(10, 6))

plt.plot(x, d(x), color='blue')
plt.plot(2, d(2), marker='o', color='blue')

plt.xlabel(r'$\mu$')
plt.ylabel(r'$d\left(\mu\right)$')
plt.grid()

plt.show()
```

```python
def g(x):
    return np.where(x >= 1, x**2, np.inf)

x = np.arange(1, 3, 0.01)

# Q1, (c)

plt.figure(figsize=(10, 6))

# The region for x < 1
plt.axvline(x=1, color='tab:red', linestyle='--', linewidth=1)
plt.fill_between([-1, 1], 0, 10, color='tab:red', alpha=0.15, label=r'$g(x) \to \infty$')

# The region for x >= 1
plt.plot(x, g(x), color='blue')
plt.plot(1, 1, marker='o', color='blue')

plt.xlim(-1, 3)
plt.ylim(0, 10)
plt.xlabel(r'$x$')
plt.ylabel(r'$g\left(x\right)$')
plt.legend()
plt.grid()

plt.show()
```

```python
# Q2, (a), (iv)

fig, axs = plt.subplots(1, 2, figsize=(10, 3))

x = np.arange(-2, 4, 0.1)

axs[0].plot(x, np.ones_like(x))
axs[0].plot(2, 1, marker='o', label=r'$\left( \mu^\star, \mathcal{L}\left( \mu^\star, x^\star \right) \right)$')
axs[0].set_xlabel(r'$\mu$')
axs[0].set_ylabel(r'$\mathcal{L}\left(\mu, x^\star\right)$')

axs[1].plot(x, (x - 1)**2 + 1)
axs[1].plot(1, 1, marker='o', label=r'$\left( x^\star, \mathcal{L}\left( \mu^\star, x^\star \right) \right)$')
axs[1].set_xlabel(r'$x$')
axs[1].set_ylabel(r'$\mathcal{L}\left(\mu^\star, x\right)$')

for ax in axs:
    ax.grid()
    ax.legend()

plt.show()
```

```python
# Q3, (b)

plt.figure(figsize=(10, 6))

mu = np.arange(0, 5, 0.1)

plt.plot(mu, L(mu, 0), label=r'$\mathcal{L}\left(\mu, 0\right)$', color='blue')
plt.plot(mu, L(mu, 1), label=r'$\mathcal{L}\left(\mu, 1\right)$', color='red')
plt.plot(mu, L(mu, 2), label=r'$\mathcal{L}\left(\mu, 2\right)$', color='green')

def envelope(mu):
    result = np.where(
        mu <= 1, mu,
        np.where(
            mu <= 3, 1,
            4 - mu
        )
    )

    return result

plt.plot(mu, envelope(mu), label='Envelope', color='magenta', linewidth=1.5)
plt.plot(mu, d(mu), label=r'$d\left(\mu\right)$', linestyle='dotted')

plt.xlabel(r'$\mu$')
plt.legend()
plt.grid()

plt.show()
```
